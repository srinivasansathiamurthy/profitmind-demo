```python
#cleanup: rename files
name_dict = {
    "ana": "Analgesics",
    "bat": "Bath Soap",
    "ber": "Beer",
    "bjc": "Bottled Juices",
    "cer": "Cereals",
    "che": "Cheeses",
    "cig": "Cigarettes",
    "coo": "Cookies",
    "cra": "Crackers",
    "cso": "Canned Soup",
    "did": "Dish Detergent",
    "fec": "Front-end-candies",
    "frd": "Frozen Dinners",
    "fre": "Frozen Entrees",
    "frj": "Frozen Juices",
    "fsf": "Fabric Softeners",
    "gro": "Grooming Products",
    "lnd": "Laundry Detergents",
    "oat": "Oatmeal",
    "ptw": "Paper Towels",
    "rfj": "Refrigerated Juices",
    "sdr": "Soft Drinks",
    "sha": "Shampoos",
    "sna": "Snack Crackers",
    "soa": "Soaps",
    "tbr": "Toothbrushes",
    "tna": "Canned Tuna",
    "tpa": "Toothpastes",
    "tti": "Bathroom Tissues",
}

#unzip files
import zipfile
from pathlib import Path

folder = Path("../data/dominicks_finer_foods")
for zip_path in folder.glob("*.zip"):
    with zipfile.ZipFile(zip_path) as zf:
        zf.extractall(folder)
    zip_path.unlink()

#i rename now.
def rename_csv(path, new_name):
    """Rename a CSV file in its current directory, keeping the .csv extension."""
    folder = os.path.dirname(path)
    new_path = os.path.join(folder, new_name if new_name.lower().endswith(".csv") else new_name + ".csv")
    if os.path.exists(new_path):
        raise FileExistsError(f"Target already exists: {new_path}")
    os.rename(path, new_path)
    return new_path


folder = "../data/dominicks_finer_foods"
for file_name in list(os.listdir("../data/dominicks_finer_foods")):
    path = os.path.join(folder, file_name)
    if file_name[0] == 'u':
        rename_csv(path, "desc_"+name_dict[file_name[3:-4]]+".csv")
    if file_name[0] == 'w':
        rename_csv(path, "data_"+name_dict[file_name[1:-4]]+".csv")
```
